```python
# =========================================
# IA DEL TITANIC
# =========================================

# Importamos librerías
import pandas as pd

# Modelo de IA
from sklearn.tree import DecisionTreeClassifier

# =========================================
# 1. LEER EL ARCHIVO CSV
# =========================================

datos = pd.read_csv("datos/titanic.csv")

# =========================================
# 2. PREPARAR LOS DATOS
# =========================================

# Elegimos columnas importantes
X = datos[["Pclass", "Sex", "Age"]].copy()

# Resultado que queremos predecir
y = datos["Survived"]

# =========================================
# 3. LIMPIAR DATOS
# =========================================

# Cambiar texto por números
# female = 0
# male = 1

X["Sex"] = X["Sex"].map({
    "female": 0,
    "male": 1
})

# Reemplazar edades vacías con promedio
promedio_edad = X["Age"].mean()

X["Age"] = X["Age"].fillna(promedio_edad)

# =========================================
# 4. CREAR LA IA
# =========================================

ia = DecisionTreeClassifier()

# Entrenar la IA
ia.fit(X, y)

print("\nIA entrenada correctamente")

# =========================================
# 5. PEDIR DATOS AL USUARIO
# =========================================

print("\n=== NUEVO PASAJERO ===")

clase = int(input("Clase del pasajero (1,2,3): "))

sexo = input("Sexo (male/female): ")

edad = int(input("Edad: "))

# Convertir texto a número
if sexo == "female":
    sexo = 0
else:
    sexo = 1

# =========================================
# 6. HACER PREDICCIÓN
# =========================================

nuevo_pasajero = pd.DataFrame({
    "Pclass": [clase],
    "Sex": [sexo],
    "Age": [edad]
})

resultado = ia.predict(nuevo_pasajero)

# =========================================
# 7. RESULTADO
# =========================================

print("\n===================================")

if resultado[0] == 1:
    print("🎉 LA IA PREDICE QUE SOBREVIVIRÍA")
else:
    print("💀 LA IA PREDICE QUE NO SOBREVIVIRÍA")

print("===================================")
```
